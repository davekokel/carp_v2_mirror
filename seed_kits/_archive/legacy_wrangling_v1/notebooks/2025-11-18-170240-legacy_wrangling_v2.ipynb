```python
# CELL 1: Setup paths and load raw inputs

import pandas as pd
from pathlib import Path

# Base directories
BASE    = Path("/Users/davekokel/Projects/carp_v2")
RAW     = BASE / "seed_kits" / "legacy_wrangling" / "raw"
WORKING = BASE / "seed_kits" / "legacy_wrangling" / "working"

# Raw input files
roi_paths_path     = RAW / "2025-11-13-092338-korra_aang_roi_root_tiffs_good-3.xlsx"
imaging_sheet_path = RAW / "2025-11-13-124226-imaging_sheet.xlsx"
inj_plasmid_path   = RAW / "Unique_injected_plasmid__preview_dqm.xlsx"
inj_rna_path       = RAW / "Unique_injected_rna__preview_dqm.xlsx"
parent_map_path    = RAW / "Unique_parent_names__mom_dad_combined__preview_dqm.xlsx"
constructs_path    = RAW / "constructs_plasmid.csv"

# Where we'll write the new annotations CSV
annot_path = WORKING / "imaging_roi_annotations_AUTO_v2.csv"

df_roi        = pd.read_excel(roi_paths_path)
df_sheet      = pd.read_excel(imaging_sheet_path)
df_inj_plasm  = pd.read_excel(inj_plasmid_path)
df_inj_rna    = pd.read_excel(inj_rna_path)
df_parent_map = pd.read_excel(parent_map_path)
df_constructs = pd.read_csv(constructs_path)

print("df_roi:", df_roi.shape)
print("df_sheet:", df_sheet.shape)
print("df_inj_plasm:", df_inj_plasm.shape)
print("df_inj_rna:", df_inj_rna.shape)
print("df_parent_map:", df_parent_map.shape)
print("df_constructs:", df_constructs.shape)

df_roi.head(), df_sheet.head()
```

```python
# CELL 2: Load flat ROI markers (one row per ROI/marker)

flat_path = WORKING / "roi_all_markers_flat_COMPLETE.csv"  # adjust if your final flat file has a different name
df_flat = pd.read_csv(flat_path)

print("df_flat:", df_flat.shape)
df_flat.head()
```

```python
# CELL 3: Define aggregators that keep ALL marker codes

def agg_uniq(series: pd.Series) -> str | None:
    """
    Aggregate all non-empty, non-NaN values into a sorted, comma-separated list.
    Returns None if there are no real values.
    """
    vals = []
    for x in series:
        if pd.isna(x):
            continue
        s = str(x).strip()
        if not s:
            continue
        vals.append(s)
    if not vals:
        return None
    uniq = sorted(set(vals))
    return ",".join(uniq)
```

```python
# CELL 4: Aggregate flat markers into per-ROI annotations (v2)

# Columns we want to aggregate as "sets" of codes
marker_cols = [
    "genotype_marker_fluor_codes",
    "genotype_marker_tag_codes",
    "treatment_plasmid_base_codes",
    "treatment_rna_base_codes",
    "treatment_marker_fluor_codes",
    "treatment_marker_tag_codes",
    "all_marker_fluor_codes",
]

# Determine ROI key and meta columns
key_col = "roi_dir"  # flat CSVs use roi_dir as the ROI key

# meta_cols should NOT include the key_col itself
meta_cols = [c for c in df_flat.columns if c not in marker_cols + [key_col]]

grouped = df_flat.groupby(key_col)

# Meta: take first row per ROI for these columns
df_meta = grouped[meta_cols].first().reset_index()  # roi_dir comes from the index

# Markers: aggregate all codes per ROI using agg_uniq
df_markers = grouped[marker_cols].agg(agg_uniq).reset_index()

# Merge back into a single per-ROI annotation dataframe
df_annot_v2 = pd.merge(df_meta, df_markers, on=key_col, how="inner")

print("df_annot_v2:", df_annot_v2.shape)
df_annot_v2.head()
```

```python
# CELL 5: Write per-ROI annotations to imaging_roi_annotations_AUTO_v2.csv

df_annot_v2.to_csv(annot_path, index=False)
print("Wrote:", annot_path, "with", len(df_annot_v2), "rows")
```
