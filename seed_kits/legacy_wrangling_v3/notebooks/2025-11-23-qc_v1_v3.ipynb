```python
# V9-QC-05 — Basecode coverage vs constructs_ft

print("V9-QC-05 — basecode coverage vs constructs")

# Load constructs & tags
constructs = pd.read_csv(CONSTRUCTS_PATH)
tags_cat = pd.read_excel(TAGS_PATH)

print("  constructs:", constructs.shape)
print("  tags_cat:  ", tags_cat.shape)

# Normalize constructs to simple [plasmid_code, fluor_code, tag_code, tag_localization]
constructs_ft = constructs.copy()
col_map = {
    "plasmid_code": "plasmid_code",
    "fluor_code": "fluor_code",
    "tag_code": "tag_code",
    "tag": "tag_code",
    "tag_position": "tag_pos",
    "tag_pos": "tag_pos",
    "tag_localization": "tag_localization",
    "localization": "tag_localization",
}
constructs_ft = constructs_ft.rename(columns={k: v for k, v in col_map.items() if k in constructs_ft.columns})

needed_cf = {"plasmid_code", "fluor_code"}
missing_cf = needed_cf - set(constructs_ft.columns)
if missing_cf:
    raise KeyError(f"constructs_ft missing required columns: {missing_cf}")

print("\nV9-QC-05 — constructs_ft columns:", list(constructs_ft.columns))

# Collect all basecodes used in df_enrich
def _split_codes(s):
    if pd.isna(s):
        return []
    vals = [t.strip() for t in str(s).split(",") for t in t.split("|")]
    return [v for v in vals if v not in ("", "None", "nan", "<NA>")]

base_cols = []
for c in ["genotype_base_codes", "treatment_rna_rna_base_code", "treatment_plasmid_plasmid_base_code"]:
    if c in df_enrich.columns:
        base_cols.append(c)

print("\nV9-QC-05 — checking basecode columns:", base_cols)

used_codes = set()
for c in base_cols:
    s = df_enrich[c].astype("string")
    for vals in s:
        for tok in _split_codes(vals):
            used_codes.add(tok)

used_codes = {c for c in used_codes if c not in ("", "None", "nan", "<NA>")}
print("  distinct basecodes in df_enrich:", len(used_codes))

# All construct basecodes (plasmids + any MGCO-type RNAs that live in constructs)
construct_codes = set(constructs_ft["plasmid_code"].astype(str).str.strip())
print("  distinct construct plasmid_codes:", len(construct_codes))

missing_codes = sorted(used_codes - construct_codes)
print("\nV9-QC-05 — basecodes used in df_enrich but NOT in constructs_ft:")
for code in missing_codes:
    print("  -", code)

if not missing_codes:
    print("  ✅ All used basecodes are present in constructs_ft.")
else:
    print("\n  ⚠️ Above basecodes need rows in constructs_plasmid.csv (with fluor_code + tag_localization).")

# Optional: show where those missing codes occur
if missing_codes:
    miss_df = (
        df_enrich
        .assign(_codes=lambda d: d[base_cols].astype(str).agg(",".join, axis=1))
    )
    mask = miss_df["_codes"].str.contains("|".join([re.escape(c) for c in missing_codes]), na=False)
    print("\nV9-05 — sample rows using missing basecodes:")
    display(
        miss_df.loc[mask, ["roi_dir", "dataset_slug"] + base_cols]
               .head(50)
    )
```

```python
# V9-08 — per-dataset QC summary

import numpy as np

if not {"dataset_slug", "all_unique_organelles"}.issubset(df_for_db.columns):
    raise KeyError("df_for_db is missing required columns for this QC cell.")

def _nonempty(s: pd.Series) -> pd.Series:
    return s.notna() & (s.astype(str).str.strip().isin(["", "None", "nan", "<NA>"]) == False)

grp = (
    df_for_db
    .assign(
        has_org=_nonempty(df_for_db["all_unique_organelles"]),
        has_geno=_nonempty(df_for_db.get("genotype_base_codes", pd.Series(index=df_for_db.index))),
        has_tx_rna=_nonempty(df_for_db.get("treatment_rna_rna_base_code", pd.Series(index=df_for_db.index))),
        has_tx_plasmid=_nonempty(df_for_db.get("treatment_plasmid_plasmid_base_code", pd.Series(index=df_for_db.index))),
    )
    .groupby("dataset_slug", dropna=False)
    .agg(
        n_rois=("roi_dir", "size"),
        n_org=("has_org", "sum"),
        n_geno=("has_geno", "sum"),
        n_tx_rna=("has_tx_rna", "sum"),
        n_tx_plasmid=("has_tx_plasmid", "sum"),
    )
    .reset_index()
)

grp["frac_org"] = grp["n_org"] / grp["n_rois"].replace(0, np.nan)
grp["frac_any_marker"] = (grp["n_geno"] + grp["n_tx_rna"] + grp["n_tx_plasmid"]) / grp["n_rois"].replace(0, np.nan)

print("V9-08 — per-dataset QC summary (sorted by frac_org asc, then n_rois desc):")
display(
    grp.sort_values(by=["frac_org", "n_rois"], ascending=[True, False]).head(40)
)

print("\nV9-08 — datasets with <100% organelle coverage (top 40):")
display(
    grp[grp["frac_org"] < 1].sort_values(by=["frac_org", "n_rois"], ascending=[True, False]).head(40)
)
```

```python
# V9-QC-EXPORT — write mapping CSVs + used constructs into qc_v9/

from pathlib import Path
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", 120)
pd.set_option("display.width", 240)

# ── 0) Paths ────────────────────────────────────────────────
ROOT = Path("/Users/davekokel/Projects/carp_v2")
BASE = ROOT / "seed_kits" / "legacy_wrangling_v3"
RAW  = BASE / "raw"
WORKING = BASE / "working"
QC_DIR = WORKING / "qc_v9"

QC_LABEL = "V9-QC-EXPORT"

QC_DIR.mkdir(parents=True, exist_ok=True)

LINKING_V5_PATH   = WORKING / "output_from_linking_v5.csv"
ENRICH_V9_PATH    = WORKING / "legacy_imaging_annotations_v9.csv"
DB_V9_PATH        = WORKING / "legacy_imaging_annotations_for_db_v9.csv"

PARENT_MAP_PATH       = RAW / "Unique_parent_names__mom_dad_combined__preview_dqm_v5.csv"
PARENT_HOLES_PATH     = RAW / "parent_hole_patch_v6.csv"
EXPT_HOLES_PATH       = RAW / "experiment_hole_patch_v7.csv"
INJECTED_RNA_PATH     = RAW / "Unique_injected_rna__preview_dqm.xlsx"
INJECTED_PLASMID_PATH = RAW / "Unique_injected_plasmid__preview_dqm.xlsx"
ORG_MAP_PATH          = RAW / "organelles_manual_mapping_v6.csv"
CONSTRUCTS_PATH       = ROOT / "seed_kits" / "2025-11-15-121231-autoload" / "constructs_plasmid.csv"

print(f"{QC_LABEL} — Paths:")
print("  ROOT:     ", ROOT)
print("  BASE:     ", BASE)
print("  WORKING:  ", WORKING)
print("  QC_DIR:   ", QC_DIR)

# ── 1) Load main tables ─────────────────────────────────────

df_struct = pd.read_csv(LINKING_V5_PATH, low_memory=False)
df_enrich = pd.read_csv(ENRICH_V9_PATH, low_memory=False)
df_for_db = pd.read_csv(DB_V9_PATH, low_memory=False)

parent_map       = pd.read_csv(PARENT_MAP_PATH)
parent_holes     = pd.read_csv(PARENT_HOLES_PATH)
exp_patch        = pd.read_csv(EXPT_HOLES_PATH)
injected_rna     = pd.read_excel(INJECTED_RNA_PATH)
injected_plasmid = pd.read_excel(INJECTED_PLASMID_PATH)
constructs_ft    = pd.read_csv(CONSTRUCTS_PATH)
org_map          = pd.read_csv(ORG_MAP_PATH)

print(f"\n{QC_LABEL} — loaded:")
print("  df_struct:      ", df_struct.shape)
print("  df_enrich:      ", df_enrich.shape)
print("  df_for_db:      ", df_for_db.shape)
print("  parent_map:     ", parent_map.shape)
print("  parent_holes:   ", parent_holes.shape)
print("  exp_patch:      ", exp_patch.shape)
print("  injected_rna:   ", injected_rna.shape)
print("  injected_plasmid:", injected_plasmid.shape)
print("  constructs_ft:  ", constructs_ft.shape)
print("  org_map:        ", org_map.shape)

# ── helpers ─────────────────────────────────────────────────

def _norm_label(s):
    if pd.isna(s):
        return None
    s = str(s).strip().lower()
    s = s.replace(" ", "")
    return s or None

def _agg_nonempty(series: pd.Series) -> str | None:
    vals = [
        str(x).strip()
        for x in series
        if pd.notna(x) and str(x).strip() not in ("", "nan", "n/a", "na")
    ]
    if not vals:
        return None
    seen = set()
    out = []
    for v in vals:
        if v not in seen:
            seen.add(v)
            out.append(v)
    return "|".join(out)

def _split_codes(val):
    if pd.isna(val):
        return []
    s = str(val).strip()
    if not s:
        return []
    return [p.strip() for p in s.replace(",", "|").split("|") if p.strip()]

def _is_empty(s: pd.Series) -> pd.Series:
    return s.isna() | (s.astype(str).str.strip().isin(["", "None", "nan", "<NA>"]))

# ── 2) parent_slug_map (from parent_map v5) ─────────────────

pm = parent_map.copy()
pm["parent_slug_norm"] = pm["parent_fish_name"].apply(_norm_label)

parent_slug_agg = (
    pm.groupby("parent_slug_norm", dropna=True)
      .agg({
          "plasmid_base_code": _agg_nonempty,
          "allele": _agg_nonempty,
          "injected_rna": _agg_nonempty,
          "injected_plasmid": _agg_nonempty,
      })
      .reset_index()
      .rename(columns={
          "plasmid_base_code": "geno_base_codes_v9_map",
          "allele": "geno_alleles_v9_map",
          "injected_rna": "inj_rna_names_v9_map",
          "injected_plasmid": "inj_plasmid_names_v9_map",
      })
)

parent_slug_agg.to_csv(QC_DIR / "qc_v9_parent_slug_map.csv", index=False)

# ── 3) parent_hole_map (from parent_hole_patch_v6) ──────────

ph = parent_holes.copy()
ph["parent_slug_norm"] = ph["parent_fish_name"].apply(_norm_label)

parent_hole_agg = (
    ph.groupby("parent_slug_norm", dropna=True)
      .agg({
          "plasmid_base_code": _agg_nonempty,
          "allele": _agg_nonempty,
          "injected_rna": _agg_nonempty,
          "injected_plasmid": _agg_nonempty,
      })
      .reset_index()
      .rename(columns={
          "plasmid_base_code": "geno_base_codes_v9_holes",
          "allele": "geno_alleles_v9_holes",
          "injected_rna": "inj_rna_names_v9_holes",
          "injected_plasmid": "inj_plasmid_names_v9_holes",
      })
)

parent_hole_agg.to_csv(QC_DIR / "qc_v9_parent_hole_map.csv", index=False)

# ── 4) experiment_slug_map (from experiment_hole_patch_v7) ─

exp = exp_patch.copy()
exp["dataset_slug_norm"] = exp["dataset_slug"].astype(str).str.strip()

exp_cols = [
    "dataset_slug",
    "genotype_base_codes",
    "genotype_allele_codes",
    "treatment_rna_base_codes",
    "treatment_plasmid_base_codes",
]
exp_cols_present = [c for c in exp_cols if c in exp.columns]

# normalize naming so it matches enrichment columns
exp = exp.rename(columns={
    "treatment_rna_base_codes": "treatment_rna_rna_base_code",
    "treatment_plasmid_base_codes": "treatment_plasmid_plasmid_base_code",
})
exp_cols_present = [
    c for c in [
        "dataset_slug",
        "genotype_base_codes",
        "genotype_allele_codes",
        "treatment_rna_rna_base_code",
        "treatment_plasmid_plasmid_base_code",
    ]
    if c in exp.columns
]

exp_slug_map = exp[exp_cols_present].drop_duplicates("dataset_slug", keep="last")
exp_slug_map.to_csv(QC_DIR / "qc_v9_experiment_slug_map.csv", index=False)

# ── 5) injected RNA / plasmid maps ─────────────────────────

rna_df = injected_rna.copy()
rna_name_col = None
rna_base_col = None
for c in rna_df.columns:
    cl = c.lower()
    if "injected_rna" in cl or cl == "rna":
        if rna_name_col is None:
            rna_name_col = c
    if "plasmid_base_code" in cl:
        if rna_base_col is None:
            rna_base_col = c

if rna_name_col and rna_base_col:
    rna_map = (
        rna_df[[rna_name_col, rna_base_col]]
        .rename(columns={rna_name_col: "treatment_name", rna_base_col: "rna_base_code"})
        .dropna(subset=["treatment_name", "rna_base_code"])
        .drop_duplicates()
    )
    rna_map["treatment_name_norm"] = rna_map["treatment_name"].apply(
        lambda s: str(s).strip() if pd.notna(s) else None
    )
    rna_map.to_csv(QC_DIR / "qc_v9_injected_rna_map.csv", index=False)
else:
    print(f"{QC_LABEL} — WARNING: could not detect injected_rna name/base columns; skipping RNA map export.")

pl_df = injected_plasmid.copy()
pl_name_col = None
pl_base_col = None
for c in pl_df.columns:
    cl = c.lower()
    if "injected_plasmid" in cl or "plasmid" in cl:
        if pl_name_col is None:
            pl_name_col = c
    if "plasmid_base_code" in cl:
        if pl_base_col is None:
            pl_base_col = c

if pl_name_col and pl_base_col:
    plasmid_map = (
        pl_df[[pl_name_col, pl_base_col]]
        .rename(columns={pl_name_col: "treatment_name", pl_base_col: "plasmid_base_code"})
        .dropna(subset=["treatment_name", "plasmid_base_code"])
        .drop_duplicates()
    )
    plasmid_map["treatment_name_norm"] = plasmid_map["treatment_name"].apply(
        lambda s: str(s).strip() if pd.notna(s) else None
    )
    plasmid_map.to_csv(QC_DIR / "qc_v9_injected_plasmid_map.csv", index=False)
else:
    print(f"{QC_LABEL} — WARNING: could not detect injected_plasmid name/base columns; skipping plasmid map export.")

# ── 6) subset of constructs_ft actually used ───────────────

used_codes = set()
for col in ["genotype_base_codes", "treatment_rna_rna_base_code", "treatment_plasmid_plasmid_base_code"]:
    if col in df_enrich.columns:
        for codes in df_enrich[col].dropna():
            for c in _split_codes(codes):
                used_codes.add(c)

constructs_used = constructs_ft[constructs_ft["plasmid_code"].isin(used_codes)].copy()

construct_cols = [c for c in ["plasmid_code", "fluor_code", "tag_code", "tag_localization"] if c in constructs_used.columns]
if construct_cols:
    constructs_used[construct_cols].drop_duplicates().to_csv(
        QC_DIR / "qc_v9_constructs_ft_used.csv", index=False
    )
else:
    pd.DataFrame(columns=["plasmid_code"]).to_csv(
        QC_DIR / "qc_v9_constructs_ft_used.csv", index=False
    )

# ── 7) organelles_manual_mapping used ──────────────────────

org_map.to_csv(QC_DIR / "qc_v9_organelles_manual_mapping_v6.csv", index=False)

# ── 8) ROI↔imaging structural mapping snapshot ─────────────

struct_cols = [
    "roi_dir",
    "dataset_slug",
    "experiment_folder",
    "roi_path_date_yyyymmdd",
    "date_mount",
    "Date imaged",
    "mount_id",
    "plate_date",
    "mount_id_inferred",
    "mount_id_source",
    "plate_id_filled",
    "slot_id_filled",
    "roi_index_within_slot",
    "bruker_roi_id",
    "link_source",
]
struct_cols = [c for c in struct_cols if c in df_struct.columns]

df_struct[struct_cols].to_csv(QC_DIR / "qc_v9_roi_struct_map.csv", index=False)

# ── 9) Global basecode usage summary ───────────────────────

usage_rows = []
for col, role in [
    ("genotype_base_codes", "genotype"),
    ("treatment_rna_rna_base_code", "rna_tx"),
    ("treatment_plasmid_plasmid_base_code", "plasmid_tx"),
]:
    if col not in df_enrich.columns:
        continue
    for codes in df_enrich[col].dropna():
        for bc in _split_codes(codes):
            usage_rows.append((bc, role))

if usage_rows:
    usage_df = pd.DataFrame(usage_rows, columns=["basecode", "role"])
    usage_pivot = (
        usage_df.pivot_table(index="basecode", columns="role", values="role", aggfunc="size", fill_value=0)
                 .reset_index()
    )

    for col in ["genotype", "rna_tx", "plasmid_tx"]:
        if col not in usage_pivot.columns:
            usage_pivot[col] = 0

    usage_pivot.rename(columns={
        "genotype":  "n_rois_genotype",
        "rna_tx":    "n_rois_rna",
        "plasmid_tx":"n_rois_plasmid",
    }, inplace=True)

    usage_pivot["used_as_genotype"]   = usage_pivot["n_rois_genotype"] > 0
    usage_pivot["used_as_rna_tx"]     = usage_pivot["n_rois_rna"] > 0
    usage_pivot["used_as_plasmid_tx"] = usage_pivot["n_rois_plasmid"] > 0

    usage_pivot.to_csv(QC_DIR / "qc_v9_basecode_usage_summary.csv", index=False)
else:
    pd.DataFrame(columns=["basecode"]).to_csv(
        QC_DIR / "qc_v9_basecode_usage_summary.csv", index=False
    )

print(f"\n{QC_LABEL} — done. QC CSVs written to {QC_DIR}")
```
