```python
# V4-02 — load parent-name catalog and construct / marker catalogs (correct paths)

RAW_V2 = ROOT / "seed_kits" / "legacy_wrangling_v2" / "raw"
STD    = ROOT / "seed_kits" / "2025-11-15-121231-autoload"

PARENT_MAP_PATH = RAW_V2 / "Unique_parent_names__mom_dad_combined__preview_dqm.xlsx"
CONSTRUCTS_PATH = STD / "constructs_plasmid.csv"
FLUORS_PATH     = STD / "fluors.csv"
TAGS_PATH       = STD / "tags.xlsx"

print("PARENT_MAP_PATH:", PARENT_MAP_PATH)
print("CONSTRUCTS_PATH:", CONSTRUCTS_PATH)
print("FLUORS_PATH:    ", FLUORS_PATH)
print("TAGS_PATH:      ", TAGS_PATH)

for p in [PARENT_MAP_PATH, CONSTRUCTS_PATH, FLUORS_PATH, TAGS_PATH]:
    if not p.exists():
        print("MISSING:", p)

# ---------------------------------------------------------------------
# Load each catalog
# ---------------------------------------------------------------------

parent_map = pd.read_excel(PARENT_MAP_PATH)
constructs = pd.read_csv(CONSTRUCTS_PATH)
fluors     = pd.read_csv(FLUORS_PATH)
tags       = pd.read_excel(TAGS_PATH)

print("\nV4-02 — parent_map shape:", parent_map.shape)
print("V4-02 — constructs shape:", constructs.shape)
print("V4-02 — fluors shape:   ", fluors.shape)
print("V4-02 — tags shape:     ", tags.shape)

print("\nV4-02 — parent_map columns:")
print(list(parent_map.columns))

print("\nV4-02 — constructs columns:")
print(list(constructs.columns))

print("\nV4-02 — fluors columns:")
print(list(fluors.columns))

print("\nV4-02 — tags columns:")
print(list(tags.columns))

print("\nV4-02 — parent_map head:")
display(parent_map.head(5))
```

```python
# V4-03 — map ZF female/male genotypes to parent_map and build genotype / treatment codes

import pandas as pd
import numpy as np
import re

df_enrich = df_struct.copy()

# ---------------------------------------------------------------------
# Check we have the ZF genotype columns
# ---------------------------------------------------------------------

expected_parent_cols = ["ZF female genotype", "ZF male genotype"]
missing_parent_cols = [c for c in expected_parent_cols if c not in df_enrich.columns]
if missing_parent_cols:
    raise KeyError(f"df_struct is missing expected ZF genotype columns: {missing_parent_cols}")

print("V4-03 — starting enrichment from ZF genotype columns")
print("  df_enrich rows:", len(df_enrich))
print("  df_enrich columns:", len(df_enrich.columns))

# ---------------------------------------------------------------------
# Normalize parent-name strings
# ---------------------------------------------------------------------

def norm_name(s):
    if pd.isna(s):
        return None
    t = str(s).strip().lower()
    t = re.sub(r"\s+", " ", t)
    return t or None

df_enrich["parent_female_name"] = df_enrich["ZF female genotype"]
df_enrich["parent_male_name"]   = df_enrich["ZF male genotype"]

df_enrich["parent_female_norm"] = df_enrich["parent_female_name"].apply(norm_name)
df_enrich["parent_male_norm"]   = df_enrich["parent_male_name"].apply(norm_name)

print("\nV4-03 — sample normalized parent names from df_enrich:")
display(
    df_enrich[
        ["parent_female_name", "parent_female_norm",
         "parent_male_name", "parent_male_norm"]
    ].drop_duplicates().head(20)
)

# ---------------------------------------------------------------------
# Normalize parent_map catalog
# ---------------------------------------------------------------------

parent_map_norm = parent_map.copy()
parent_map_norm["parent_name_norm"] = parent_map_norm["parent_fish_name"].apply(norm_name)

pm_cols = [
    "parent_name_norm",
    "plasmid_base_code",
    "allele",
    "injected_rna",
    "injected_plasmid",
]
pm_small = parent_map_norm[pm_cols].copy()

# ---------------------------------------------------------------------
# Pipe-splitting helpers
# ---------------------------------------------------------------------

def split_pipe(val):
    if val is None or (isinstance(val, float) and pd.isna(val)):
        return []
    s = str(val).strip()
    if not s:
        return []
    return [x.strip() for x in s.split("|") if x.strip()]

def join_unique(items):
    out = []
    seen = set()
    for x in items:
        if x is None:
            continue
        sx = str(x).strip()
        if not sx:
            continue
        if sx not in seen:
            seen.add(sx)
            out.append(sx)
    return "|".join(out) if out else None

# ---------------------------------------------------------------------
# Merge parent_map for female and male parents
# ---------------------------------------------------------------------

df_enrich = df_enrich.merge(
    pm_small.add_prefix("female_"),
    left_on="parent_female_norm",
    right_on="female_parent_name_norm",
    how="left",
)

df_enrich = df_enrich.merge(
    pm_small.add_prefix("male_"),
    left_on="parent_male_norm",
    right_on="male_parent_name_norm",
    how="left",
)

print("\nV4-03 — after merging parent_map:")
print("  columns:", len(df_enrich.columns))

# ---------------------------------------------------------------------
# Build genotype_base_codes and genotype_allele_codes
# ---------------------------------------------------------------------

def combine_parent_codes(row, which):
    """
    which = 'plasmid' or 'allele'
    """
    if which == "plasmid":
        fem = row.get("female_plasmid_base_code")
        mal = row.get("male_plasmid_base_code")
    elif which == "allele":
        fem = row.get("female_allele")
        mal = row.get("male_allele")
    else:
        return None

    items = split_pipe(fem) + split_pipe(mal)
    return join_unique(items)

df_enrich["genotype_base_codes"]   = df_enrich.apply(lambda r: combine_parent_codes(r, "plasmid"), axis=1)
df_enrich["genotype_allele_codes"] = df_enrich.apply(lambda r: combine_parent_codes(r, "allele"), axis=1)

# ---------------------------------------------------------------------
# Build treatment_plasmid_base_codes / treatment_rna_base_codes
# ---------------------------------------------------------------------

def combine_treatment_codes(row, which):
    if which == "plasmid":
        fem = row.get("female_injected_plasmid")
        mal = row.get("male_injected_plasmid")
    elif which == "rna":
        fem = row.get("female_injected_rna")
        mal = row.get("male_injected_rna")
    else:
        return None

    items = split_pipe(fem) + split_pipe(mal)
    return join_unique(items)

df_enrich["treatment_plasmid_base_codes"] = df_enrich.apply(
    lambda r: combine_treatment_codes(r, "plasmid"), axis=1
)
df_enrich["treatment_rna_base_codes"] = df_enrich.apply(
    lambda r: combine_treatment_codes(r, "rna"), axis=1
)

# ---------------------------------------------------------------------
# QC: coverage and sample rows
# ---------------------------------------------------------------------

has_geno = df_enrich["genotype_base_codes"].notna()
n_with_geno = has_geno.sum()

print("\nV4-03 — genotype coverage:")
print("  rows with genotype_base_codes:", n_with_geno, "/", len(df_enrich))

print("\nV4-03 — sample rows with genotype_base_codes:")
display(
    df_enrich.loc[has_geno, [
        "roi_dir",
        "parent_female_name",
        "parent_male_name",
        "genotype_base_codes",
        "genotype_allele_codes",
        "treatment_plasmid_base_codes",
        "treatment_rna_base_codes",
    ]].head(20)
)

print("\nV4-03 — sample rows with NO genotype_base_codes (for inspection):")
display(
    df_enrich.loc[~has_geno, [
        "roi_dir",
        "parent_female_name",
        "parent_male_name",
    ]].head(20)
)
```

```python
# V4-04 — attach constructs/fluors/tags and build genotype marker/localization columns

import pandas as pd
import numpy as np

print("V4-04 — starting from df_enrich shape:", df_enrich.shape)

# ---------------------------------------------------------------------
# 1) Normalize constructs, fluors, tags
# ---------------------------------------------------------------------

# constructs: we have plasmid_code, fluor_code, tag_code, tag_pos
constructs_small = constructs[[
    "plasmid_code",
    "fluor_code",
    "tag_code",
    "tag_pos",
]].copy()

print("V4-04 — constructs_small sample:")
display(constructs_small.head(10))

# fluors.csv: columns = nickname, excitation_nm, emission_nm, note, aliases
# There is no explicit fluor_code column; we assume constructs.fluor_code == fluors.nickname
fluors_small = fluors.rename(columns={"nickname": "fluor_code"})
fluors_small = fluors_small[[
    "fluor_code",
    "excitation_nm",
    "emission_nm",
    "note",
    "aliases",
]].copy()

print("\nV4-04 — fluors_small sample:")
display(fluors_small.head(10))

# tags.xlsx: columns = nickname, localization, note, citation_link, aliases
# Assume constructs.tag_code == tags.nickname
tags_small = tags.rename(columns={"nickname": "tag_code", "localization": "tag_localization"})
tags_small = tags_small[[
    "tag_code",
    "tag_localization",
    "note",
    "citation_link",
    "aliases",
]].copy()

print("\nV4-04 — tags_small sample:")
display(tags_small.head(10))

# QC: fluor_code coverage
fluor_codes_constructs = set(constructs_small["fluor_code"].dropna().unique())
fluor_codes_catalog    = set(fluors_small["fluor_code"].dropna().unique())
missing_fluors = sorted(fluor_codes_constructs - fluor_codes_catalog)
print("\nV4-04 — unique fluor_codes in constructs:", len(fluor_codes_constructs))
print("V4-04 — unique fluor_codes in fluors:    ", len(fluor_codes_catalog))
print("V4-04 — fluor_codes in constructs but not in fluors:", missing_fluors[:20])

# QC: tag_code coverage
tag_codes_constructs = set(constructs_small["tag_code"].dropna().unique())
tag_codes_catalog    = set(tags_small["tag_code"].dropna().unique())
missing_tags = sorted(tag_codes_constructs - tag_codes_catalog)
print("\nV4-04 — unique tag_codes in constructs:", len(tag_codes_constructs))
print("V4-04 — unique tag_codes in tags:      ", len(tag_codes_catalog))
print("V4-04 — tag_codes in constructs but not in tags:", missing_tags[:20])

# ---------------------------------------------------------------------
# 2) Build per-plasmid marker metadata
# ---------------------------------------------------------------------

# Join constructs -> fluors
constructs_f = constructs_small.merge(
    fluors_small,
    on="fluor_code",
    how="left",
    suffixes=("", "_fluor"),
)

# Join constructs_f -> tags
constructs_ft = constructs_f.merge(
    tags_small,
    on="tag_code",
    how="left",
    suffixes=("", "_tag"),
)

print("\nV4-04 — constructs_ft (per plasmid marker info) sample:")
display(constructs_ft.head(15))

# ---------------------------------------------------------------------
# 3) Expand genotype_base_codes per ROI into rows and join marker info
# ---------------------------------------------------------------------

# Only rows with genotype_base_codes
geno = df_enrich[["roi_dir", "genotype_base_codes"]].dropna().copy()
geno["plasmid_code"] = geno["genotype_base_codes"].apply(lambda s: split_pipe(s))
geno = geno.explode("plasmid_code").reset_index(drop=True)

print("\nV4-04 — geno (exploded genotype_base_codes) sample:")
display(geno.head(15))

# Join with constructs_ft to attach fluor, tag, localization per (ROI, plasmid)
geno_markers = geno.merge(
    constructs_ft,
    on="plasmid_code",
    how="left",
)

print("\nV4-04 — geno_markers sample:")
display(
    geno_markers[
        ["roi_dir", "plasmid_code", "fluor_code", "tag_code", "tag_localization", "tag_pos"]
    ].head(20)
)

# ---------------------------------------------------------------------
# 4) Aggregate marker info back to per-ROI columns
# ---------------------------------------------------------------------

def agg_marker_lists(subdf):
    fluor_codes = join_unique(subdf["fluor_code"].tolist())
    tag_codes   = join_unique(subdf["tag_code"].tolist())
    locs        = join_unique(subdf["tag_localization"].tolist())

    # fusion label: "fluor_code::tag_code"
    # if either side is missing, skip that combo
    fusion_labels = []
    for fc, tc in zip(subdf["fluor_code"], subdf["tag_code"]):
        if pd.isna(fc) or pd.isna(tc):
            continue
        fusion_labels.append(f"{fc}::{tc}")
    fusion_labels = join_unique(fusion_labels)

    # fluor(localization) combos: "fluor_code(localization)"
    fluor_org = []
    for fc, loc in zip(subdf["fluor_code"], subdf["tag_localization"]):
        if pd.isna(fc) or pd.isna(loc):
            continue
        fluor_org.append(f"{fc}({loc})")
    fluor_org = join_unique(fluor_org)

    return pd.Series({
        "genotype_marker_fluor_codes": fluor_codes,
        "genotype_marker_tag_codes": tag_codes,
        "genotype_marker_localizations": locs,
        "genotype_marker_fusion_labels": fusion_labels,
        "all_unique_organelles": locs,          # for now: genotype-only organelles
        "all_fluor_organelles": fluor_org,      # genotype-only for now
    })

geno_agg = (
    geno_markers
    .groupby("roi_dir", dropna=False)
    .apply(agg_marker_lists)
    .reset_index()
)

print("\nV4-04 — geno_agg (per-ROI marker rollup) sample:")
display(geno_agg.head(20))

# ---------------------------------------------------------------------
# 5) Merge marker rollup back into df_enrich
# ---------------------------------------------------------------------

before_cols = set(df_enrich.columns)

df_enrich = df_enrich.merge(
    geno_agg,
    on="roi_dir",
    how="left",
)

after_cols = set(df_enrich.columns)
new_cols = sorted(after_cols - before_cols)

print("\nV4-04 — df_enrich after merging marker rollup:")
print("  shape:", df_enrich.shape)
print("  new marker columns:", new_cols)

# QC: sample enriched rows
print("\nV4-04 — sample enriched rows (with marker columns):")
display(
    df_enrich.loc[
        df_enrich["genotype_marker_fluor_codes"].notna(),
        [
            "roi_dir",
            "genotype_base_codes",
            "genotype_marker_fluor_codes",
            "genotype_marker_tag_codes",
            "genotype_marker_localizations",
            "genotype_marker_fusion_labels",
            "all_unique_organelles",
            "all_fluor_organelles",
        ],
    ].head(20)
)
```

```python
# V4-04b — assign default "cytosol" tag/localization to untagged genotype markers

import numpy as np
import pandas as pd

print("V4-04b — initial organelle coverage (before cytosol default):")
print(df_enrich["all_unique_organelles"].notna().value_counts())

# mask: ROIs with genotype fluor markers but no tag/localization yet
mask_untagged = (
    df_enrich["genotype_marker_fluor_codes"].notna()
    & df_enrich["genotype_marker_tag_codes"].isna()
    & df_enrich["genotype_marker_localizations"].isna()
)

n_untagged = mask_untagged.sum()
print("V4-04b — ROIs with fluor but no tag/localization:", n_untagged)

# 1) set default tag/localization
df_enrich.loc[mask_untagged, "genotype_marker_tag_codes"] = "cytosol"
df_enrich.loc[mask_untagged, "genotype_marker_localizations"] = "cytosol"

# 2) ensure all_unique_organelles includes 'cytosol'
def add_cytosol(orgs):
    if pd.isna(orgs) or not str(orgs).strip():
        return "cytosol"
    toks = [t.strip() for t in str(orgs).split("|") if t.strip()]
    if "cytosol" in toks:
        return orgs
    return orgs + "|cytosol"

df_enrich.loc[mask_untagged, "all_unique_organelles"] = df_enrich.loc[
    mask_untagged, "all_unique_organelles"
].apply(add_cytosol)

# 3) ensure all_fluor_organelles includes fluor_code(cytosol)
def add_cytosol_fluor_org(row):
    org = row["all_fluor_organelles"]
    fl  = row["genotype_marker_fluor_codes"]
    if pd.isna(fl) or not str(fl).strip():
        return org

    # genotype_marker_fluor_codes may be pipe-separated
    fcodes = [t.strip() for t in str(fl).split("|") if t.strip()]
    new_tokens = [f"{fc}(cytosol)" for fc in fcodes]

    if pd.isna(org) or not str(org).strip():
        return "|".join(new_tokens)

    toks = [t.strip() for t in str(org).split("|") if t.strip()]
    existing = set(toks)
    for nt in new_tokens:
        if nt not in existing:
            toks.append(nt)
    return "|".join(toks)

df_enrich.loc[mask_untagged, "all_fluor_organelles"] = df_enrich.loc[
    mask_untagged
].apply(add_cytosol_fluor_org, axis=1)

print("\nV4-04b — organelle coverage AFTER cytosol default:")
print(df_enrich["all_unique_organelles"].notna().value_counts())

print("\nV4-04b — sample untagged (now cytosol) rows:")
display(
    df_enrich.loc[mask_untagged, [
        "roi_dir",
        "genotype_marker_fluor_codes",
        "genotype_marker_tag_codes",
        "genotype_marker_localizations",
        "all_unique_organelles",
        "all_fluor_organelles",
    ]].head(20)
)
```

```python
# V4-04c — merge "additional injected" treatments from imaging_sheet without duplicating ROIs
#          and rename them to treatment_*_names_sheet

from pathlib import Path
import pandas as pd

RAW_V2 = ROOT / "seed_kits" / "legacy_wrangling_v2" / "raw"
IMAGING_SHEET_FILE = RAW_V2 / "2025-11-21-220012-imaging_sheet.xlsx"

print("V4-04c — imaging sheet file:", IMAGING_SHEET_FILE)
if not IMAGING_SHEET_FILE.exists():
    raise FileNotFoundError(f"Imaging sheet not found at {IMAGING_SHEET_FILE}")

df_imaging_sheet = pd.read_excel(IMAGING_SHEET_FILE)

print("V4-04c — df_imaging_sheet shape:", df_imaging_sheet.shape)
print("V4-04c — df_imaging_sheet columns:")
print(list(df_imaging_sheet.columns))

# Column names from the sheet (typos and all)
COL_DATA_LOCATION         = "Data location"
COL_ADD_PLASMIDS_INJECTED = "additional plasmids injected"
COL_ADD_MRNAS_INJECTED    = "additional mRNAs injected"
COL_ADD_PROTEINS_INJECTED = "additonal proteins injected"
COL_ADD_DYES_INJECTED     = "additonal dye and chemicals"

needed_cols = [
    COL_DATA_LOCATION,
    COL_ADD_PLASMIDS_INJECTED,
    COL_ADD_MRNAS_INJECTED,
    COL_ADD_PROTEINS_INJECTED,
    COL_ADD_DYES_INJECTED,
]

missing = [c for c in needed_cols if c not in df_imaging_sheet.columns]
if missing:
    print("V4-04c — WARNING: missing expected columns in imaging sheet:", missing)
    needed_cols = [c for c in needed_cols if c in df_imaging_sheet.columns]

sheet_small = df_imaging_sheet[needed_cols].copy()

def norm_path_str(s):
    if pd.isna(s):
        return None
    return str(s).strip().replace("\\", "/")

sheet_small["data_location_norm"] = sheet_small[COL_DATA_LOCATION].apply(norm_path_str)
df_enrich["data_location_norm"]   = df_enrich["Data location"].apply(norm_path_str)

print("\nV4-04c — nunique data_location_norm in sheet:", sheet_small["data_location_norm"].nunique())
print("V4-04c — nunique data_location_norm in df_enrich:", df_enrich["data_location_norm"].nunique())

# Aggregate to ONE ROW per data_location_norm so we don't duplicate ROIs
def agg_text(col):
    vals = [str(x).strip() for x in col.dropna() if str(x).strip()]
    if not vals:
        return None
    out = []
    seen = set()
    for v in vals:
        if v not in seen:
            seen.add(v)
            out.append(v)
    return "|".join(out)

group_cols = ["data_location_norm"]
agg_dict = {}

if COL_ADD_PLASMIDS_INJECTED in sheet_small.columns:
    agg_dict[COL_ADD_PLASMIDS_INJECTED] = agg_text
if COL_ADD_MRNAS_INJECTED in sheet_small.columns:
    agg_dict[COL_ADD_MRNAS_INJECTED]    = agg_text
if COL_ADD_PROTEINS_INJECTED in sheet_small.columns:
    agg_dict[COL_ADD_PROTEINS_INJECTED] = agg_text
if COL_ADD_DYES_INJECTED in sheet_small.columns:
    agg_dict[COL_ADD_DYES_INJECTED]     = agg_text

sheet_agg = (
    sheet_small
    .groupby(group_cols, dropna=True)
    .agg(agg_dict)
    .reset_index()
)

print("\nV4-04c — sheet_agg shape:", sheet_agg.shape)
print("V4-04c — sheet_agg sample:")
display(sheet_agg.head(20))

before_rows = len(df_enrich)
before_roi_unique = df_enrich["roi_dir"].nunique()
before_cols = set(df_enrich.columns)

df_enrich = df_enrich.merge(
    sheet_agg,
    on="data_location_norm",
    how="left",
    suffixes=("", "_sheet"),
)

after_rows = len(df_enrich)
after_roi_unique = df_enrich["roi_dir"].nunique()
after_cols = set(df_enrich.columns)

print("\nV4-04c — df_enrich after merge:")
print("  rows (before):", before_rows, "  rows (after):", after_rows)
print("  unique roi_dir (before):", before_roi_unique, "  (after):", after_roi_unique)
print("  new columns added:", sorted(after_cols - before_cols))

# Rename for clarity
rename_map = {}
if COL_ADD_PLASMIDS_INJECTED in df_enrich.columns:
    rename_map[COL_ADD_PLASMIDS_INJECTED] = "treatment_plasmid_names_sheet"
if COL_ADD_MRNAS_INJECTED in df_enrich.columns:
    rename_map[COL_ADD_MRNAS_INJECTED]    = "treatment_rna_names_sheet"
if COL_ADD_PROTEINS_INJECTED in df_enrich.columns:
    rename_map[COL_ADD_PROTEINS_INJECTED] = "treatment_protein_names_sheet"
if COL_ADD_DYES_INJECTED in df_enrich.columns:
    rename_map[COL_ADD_DYES_INJECTED]     = "treatment_dye_names_sheet"

df_enrich = df_enrich.rename(columns=rename_map)

print("\nV4-04c — sample enriched treatments from imaging sheet:")
cols_show = ["roi_dir", "Data location"]
cols_show += [c for c in [
    "treatment_plasmid_names_sheet",
    "treatment_rna_names_sheet",
    "treatment_protein_names_sheet",
    "treatment_dye_names_sheet",
] if c in df_enrich.columns]

display(df_enrich[cols_show].head(20))

# Also inspect the specific ROI we care about:
mask_0428 = df_enrich["roi_dir"].str.contains("20250428_mem_histone/fish1_72hpf/roi1_tail")
print("\nV4-04c — specific ROI row for 20250428_mem_histone/fish1_72hpf/roi1_tail:")
display(df_enrich.loc[mask_0428, cols_show])
```

```python
# V4-05 — build final v4 outputs and write CSVs (with treatment-name columns)

import pandas as pd

print("V4-05 — starting from df_enrich shape:", df_enrich.shape)

# ---------------------------------------------------------------------
# 1) Basic QC: row counts and duplicates by roi_dir
# ---------------------------------------------------------------------

n_rows = len(df_enrich)
n_roi_unique = df_enrich["roi_dir"].nunique()
print("V4-05 — df_enrich rows:", n_rows)
print("V4-05 — unique roi_dir:", n_roi_unique)

dups = df_enrich[df_enrich["roi_dir"].duplicated(keep=False)]
print("V4-05 — duplicate roi_dir rows:", len(dups))
if len(dups):
    print("V4-05 — sample duplicates:")
    display(dups[["roi_dir", "bruker_roi_id"]].head(20))

# ---------------------------------------------------------------------
# 2) Define final column sets
# ---------------------------------------------------------------------

base_for_db_cols = [
    # IDs / structural
    "roi_dir",
    "bruker_roi_id",
    "plate_date",
    "plate_id_filled",
    "slot_id_filled",
    "roi_index_within_slot",
    "dataset_slug",
    "experiment_folder",
    "fish_id",
    "fish_number",
    "fish_age_hpf",
    "roi_anatomy",
    "roi_tiffs",

    # imaging timing
    "date_experiment",
    "Date imaged",
    "date_mount",

    # ZF genotypes (raw from sheet)
    "ZF female genotype",
    "ZF male genotype",

    # parent-map-derived genotype
    "genotype_base_codes",
    "genotype_allele_codes",

    # treatment codes (from parent_map)
    "treatment_plasmid_base_codes",
    "treatment_rna_base_codes",

    # genotype markers and organelles
    "genotype_marker_fluor_codes",
    "genotype_marker_tag_codes",
    "genotype_marker_localizations",
    "genotype_marker_fusion_labels",
    "all_unique_organelles",
    "all_fluor_organelles",

    # provenance
    "link_source",
]

# imaging-sheet treatment names (if present)
sheet_treatment_cols = [
    "treatment_plasmid_names_sheet",
    "treatment_rna_names_sheet",
    "treatment_protein_names_sheet",
    "treatment_dye_names_sheet",
]

for_db_cols = base_for_db_cols + sheet_treatment_cols

missing_for_db = [c for c in for_db_cols if c not in df_enrich.columns]
if missing_for_db:
    print("V4-05 — WARNING: these requested for_db_cols are missing and will be dropped:", missing_for_db)
    for_db_cols = [c for c in for_db_cols if c in df_enrich.columns]

df_for_db = df_enrich[for_db_cols].copy()

print("\nV4-05 — df_for_db shape:", df_for_db.shape)
print("V4-05 — df_for_db columns:")
print(for_db_cols)

# ---------------------------------------------------------------------
# 3) Write outputs
# ---------------------------------------------------------------------

OUT_DIR = ROOT / "seed_kits" / "legacy_wrangling_v2" / "working"

FULL_OUT = OUT_DIR / "legacy_imaging_annotations_v4.csv"
DB_OUT   = OUT_DIR / "legacy_imaging_annotations_for_db_v4.csv"

df_enrich.to_csv(FULL_OUT, index=False)
df_for_db.to_csv(DB_OUT, index=False)

print("\nV4-05 — wrote full annotations to:", FULL_OUT)
print("V4-05 — wrote DB subset to:", DB_OUT)

# ---------------------------------------------------------------------
# 4) Quick summary of DB subset for sanity
# ---------------------------------------------------------------------

print("\nV4-05 — DB subset sample:")
display(df_for_db.head(20))

print("\nV4-05 — DB subset organelle coverage:")
print(
    df_for_db["all_unique_organelles"].notna().value_counts()
)
```
